```python
from pathlib import Path

import numpy as np

from matplotlib import pyplot as plt
```

```python
# Fully-qualified path to labeled training file to edit.
ltName: Path = Path('/explore/nobackup/people/rlgill/SystemTesting/modis-vcf/MOD44/h10v06.bin')

# Upper-left pixel location of the area to edit
ul: (int, int) = (1500, 2500)

# Width in pixels of the rectangle to edit
width: int = 200

# Height in pixels of the rectangle to edit
height: int = 2000

# Constant value with which to replace all pixels in the rectangle
NEW_VALUE: int = 8
```

```python
rect: np.ndarray = np.fromfile(ltName, dtype=np.uint8).reshape(4800, 4800)

print('Current')
print('Data type:', rect.dtype)
print('Min:', rect.min())
print('Max:', rect.max())
print('Mean:', rect.mean())

rect[ul[0]:ul[0]+height+1, ul[1]:ul[1]+width+1] = NEW_VALUE

print('New')
print('Data type:', rect.dtype)
print('Min:', rect.min())
print('Max:', rect.max())
print('Mean:', rect.mean())

plt.imshow(rect, interpolation='nearest')
plt.show()

print('Uncomment the following line to overwrite', ltName.name)
# rect.tofile(ltName)  # Uncomment to write the file to disk.
```
